```python
for k, v in PARSE_DICT.items():
    values = df[k].values
    expanded = list(map( lambda x: parse_bit_field(x, v), values ) )
    expanded_df = pd.DataFrame(expanded)

    df = pd.concat( (df, expanded_df ), axis=1 )
```

```python
for c in df.columns[1:]:
    fig, ax= plt.subplots()
    # df.plot(x="Time", y="TG Rotate")
    ax.plot(df['Time'].values, df[c].values)

    ax.xaxis.set_major_formatter(matplotlib.dates.DateFormatter('%Y-%m-%d %H:%M:%S'))    
    ax.set_title(c)
    fig.autofmt_xdate()
    # break
    name = c.replace("/", "-")
    fig.savefig(f"data/chamber_log2_vis/{name}.png")
    plt.close(fig)
```
