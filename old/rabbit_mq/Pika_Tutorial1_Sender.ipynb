```python

#!/usr/bin/env python
import pika
```

```python
connection = pika.BlockingConnection(pika.ConnectionParameters('localhost'))
channel = connection.channel()
```

```python
# create a queue call hello in RabbitMQ
channel.queue_declare(queue='hello')

# can use 
# sudo docker exec -it rabbitmq rabbitmqctl list_queues
# to check the queue in docker image
```

```python
# routing key is used to matched with the queue name, which is hello
channel.basic_publish(exchange='',
                      routing_key='hello',
                      body='Hello World!')
print(" [x] Sent 'Hello World!'")
```

```python
import time
for i in range(100):
    time.sleep(2)
    msg = f'Hello World! ({i})'
    channel.basic_publish(exchange='',
                          routing_key='hello',
                          body=msg)
    print(f"[x] {msg}")
```

```python
connection.close()
```
