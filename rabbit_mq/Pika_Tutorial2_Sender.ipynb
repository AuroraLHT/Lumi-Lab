```python

#!/usr/bin/env python
import pika
```

```python
connection = pika.BlockingConnection(pika.ConnectionParameters('localhost'))
channel = connection.channel()
```

```python
# create a queue call hello in RabbitMQ
channel.queue_declare(queue='worker')

# can use 
# sudo docker exec -it rabbitmq rabbitmqctl list_queues
# to check the queue in docker image
```

```python
import time
for i in range(1, 5):
    time.sleep(2)
    msg = f'This task take {i} sec to complete: tasklength[' + i*"."+ ']'
    channel.basic_publish(exchange='',
                          routing_key='worker',
                          body=msg)
    print(f" [x] {msg}")
```

```python
connection.close()
```
